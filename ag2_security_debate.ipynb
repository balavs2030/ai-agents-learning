Import Libraries

```python
import os
from dotenv import load_dotenv
from autogen import ConversableAgent

load_dotenv()
print("Imports successful!")
```

Configure the LLM

```python
llm_config = {
    "config_list": [
        {
            "model": "gpt-4o-mini",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    ],
    "temperature": 0,
}

print("LLM config ready!")
```

Define two agents

```python
# Security Architect — proposes the policy
architect = ConversableAgent(
    name="Security_Architect",
    system_message="""You are a senior cloud security architect with 20 years 
    of experience across Azure, AWS and GCP. You propose comprehensive security 
    policies for cloud infrastructure. When you are satisfied the policy is 
    complete and all concerns addressed, say TERMINATE.""",
    human_input_mode="NEVER",
    llm_config=llm_config,
    max_consecutive_auto_reply=3,
)

# Security Critic — challenges and finds gaps
critic = ConversableAgent(
    name="Security_Critic",
    system_message="""You are a security auditor and red team expert. 
    You critically review security policies looking for gaps, weaknesses, 
    and missing controls using OWASP and CIS Benchmarks. 
    Ask specific challenging questions to improve the policy.
    When you are satisfied the policy is comprehensive, say TERMINATE.""",
    human_input_mode="NEVER",
    llm_config=llm_config,
    max_consecutive_auto_reply=3,
)

print("Agents defined!")
```

Start the debate

```python
# Critic initiates by challenging the architect
chat_result = critic.initiate_chat(
    recipient=architect,
    message="""Please propose a security policy for a Kubernetes cluster 
    deployed on Azure in a production environment handling sensitive data.""",
    max_turns=4,
    summary_method="reflection_with_llm"
)
```

Printing Debate Summary

```python
print("\n" + "="*60)
print("DEBATE SUMMARY:")
print("="*60)
print(chat_result.summary)
```
