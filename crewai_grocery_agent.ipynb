Import Libraries

```python
from crewai import Agent, Task, Crew, Process
from crewai import LLM
import os
from dotenv import load_dotenv

load_dotenv()
print("Imports Successful!!!")
```

Initialize LLM

```python
llm = LLM(
    model = "anthropic/claude-sonnet-4-6",
    api_key = os.getenv("ANTHROPIC_API_KEY")
)
print("LLM Ready!!!")
```

Chef Agent

```python
chef_agent = Agent(
    role = "Home cooking chef.",
    goal = "Help users pick healthy ingredients and create awesome meals that are quick, easy and tasty.",
    backstory = "A top class home cooking chef whose style matches mom's cooking — personal, emotional, and deeply satisfying. Known for making healthy food feel like comfort food.",
    verbose=True,
    allow_delegation=False,
    llm = llm
)
print("Chef Agent ready!!!")
```

Dr Ganapathy Agent

```python
dr_ganapathy_agent = Agent(
    role="Nutritionist and Health Advisor",
    goal="Review proposed meal plans and ingredients critically to ensure they align with the patient's dietary needs and health goals.",
    backstory="An industry-known health advisor — Dr. Advice Ganapathy — who evaluates the health quotient of every meal, flags concerns, and suggests specific healthier alternatives where needed.",
    verbose=True,
    allow_delegation=False,
    llm=llm
)
print("Dr. Ganapathy Agent ready!!!")
```

Research Agent

Grocery Matcher Agent

```python
grocery_matcher_agent = Agent(
    role="Grocery Identifier",
    goal="List out the ingredients from the available store inventory matching the meal plan requirements.",
    backstory="A grocery specialist who searches the store inventory and lists available items as-is or suggests suitable substitutes when an ingredient is unavailable.",
    verbose=True,
    allow_delegation=False,
    llm=llm
)
print("Grocery Matcher Agent ready!!!")
```

Chef Task

```python
chef_task = Task(
    description="""Create a healthy meal plan based on the user's dietary needs: {dietary_needs}.
    Suggest a complete meal with a brief recipe and list all specific ingredients 
    needed at the end in a clear numbered list.""",
    expected_output="""A complete meal recommendation including:
    - Meal name and brief description
    - Simple cooking instructions
    - Numbered list of all ingredients with quantities""",
    agent=chef_agent
)
print("Chef Task defined!!!")
```

Dr Ganapathy Task

```python
dr_ganapathy_task = Task(
    description="""Review the proposed meal plan and ingredients critically based on 
    the user's dietary needs: {dietary_needs}.
    Check nutritional value, health quotient, and meal timing appropriateness.
    If the meal is nutritionally sound say 'APPROVED', otherwise suggest improvements.""",
    expected_output="""A health assessment including:
    - Nutritional strengths and weaknesses
    - Health quotient score
    - Specific improvements or healthier alternatives
    - Final verdict: APPROVED or needs revision""",
    agent=dr_ganapathy_agent
)
print("Dr. Ganapathy Task defined!")
```

Research Task

Grocery Matcher Task

```python
grocery_matcher_task = Task(
    description="""Based on the meal plan and dietary recommendations provided, 
    identify and list all required groceries from the available store inventory.""",
    expected_output="""A complete grocery list including:
    - All ingredients available in the inventory
    - Any missing ingredients highlighted clearly
    - Alternative substitutes suggested for unavailable items""",
    agent=grocery_matcher_agent
)
print("Grocery Matcher Task defined!!!")
```

Assemble the Crew

```python
crew = Crew(
    agents=[chef_agent, dr_ganapathy_agent, grocery_matcher_agent],
    tasks=[chef_task, dr_ganapathy_task, grocery_matcher_task],
    process=Process.sequential,
    verbose=True
)
print("Crew assembled!!!")
```

Run the Crew

```python
import asyncio

async def run_crew():
    result = await crew.kickoff_async(
        inputs={"dietary_needs": "Good Vegeterian Carb meal for junior tennis athlete for dinner. Eggs are ok"}
    )
    return result

result = await run_crew()
print("\n Final Result:")
print(result)
```
